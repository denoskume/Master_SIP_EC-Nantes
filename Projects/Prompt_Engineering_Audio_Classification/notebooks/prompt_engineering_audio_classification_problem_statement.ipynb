## Scope and Constraints

The experiment remains strictly zero-shot with respect to ESC-50: no ESC-50-specific classifier training or fine-tuning is performed. Results depend on the selected pretrained CLAP checkpoint, software environment, prompt set, and evaluation protocol. Published external benchmark values are contextual references rather than a controlled head-to-head comparison.
