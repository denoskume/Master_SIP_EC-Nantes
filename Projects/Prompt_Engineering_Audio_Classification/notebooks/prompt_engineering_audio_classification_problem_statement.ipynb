## Context

Contrastive Language-Audio Pretraining (CLAP) maps audio signals and natural-language descriptions into a shared embedding space. This makes it possible to classify an unseen audio sample without training a task-specific classifier: candidate class names are converted into text prompts, embedded by the CLAP text encoder, and ranked against the audio embedding by cross-modal similarity.

In this project, prompt formulation is treated as an experimental variable. The same pretrained CLAP checkpoint, ESC-50 evaluation set, candidate class vocabulary, embedding normalization, and similarity rule are retained while only the surrounding text template changes.

The technical challenge is therefore not only to obtain a high classification score, but to design a controlled zero-shot experiment capable of showing whether changes in accuracy are attributable to prompt wording rather than to changes in model, data, or evaluation procedure.

## Problem Statement

Evaluate how natural-language prompt formulation influences CLAP zero-shot environmental-sound classification on the ESC-50 dataset.

The study must:

- prepare and validate the complete ESC-50 evaluation dataset;
- initialize one fixed pretrained LAION-CLAP model for all experiments;
- define a controlled set of ten prompt strategies including a class-only baseline;
- compute normalized audio and text embeddings using one common inference pipeline;
- rank the 50 candidate ESC-50 classes by audio-text similarity;
- benchmark every prompt strategy on the same 2,000 audio samples;
- identify the strongest measured prompt and quantify its gain over the class-only baseline;
- analyze class-wise and Top-k behavior;
- investigate whether class-wise accuracy is associated with selected AudioSet metadata variables;
- contextualize the zero-shot result against recorded published ESC-50 classifiers without treating unlike training protocols as directly equivalent;
- generate reproducible tables, figures, prediction records, and a final summary derived from computed results.

ESC-50-specific training, fine-tuning, learned prompt optimization, and supervised classifier-head training are outside the defined scope.

## 6. Zero-Shot Classification Framework

Implement one common inference procedure for all prompt strategies.

For each template:

1. instantiate one text prompt for each of the 50 classes;
2. compute CLAP text embeddings;
3. compute the CLAP audio embedding for each ESC-50 sample;
4. L2-normalize both modalities;
5. compute the similarity vector over all candidate classes;
6. rank the 50 classes;
7. record Top-1 prediction, correctness, similarity scores, confidence margin, and Top-k class ranking.

For normalized audio embedding $a$ and text embedding $t_c$,

$$
s_c=a^Tt_c
$$

is equivalent to cosine similarity.

The predicted class is

$$
\hat c=
\operatorname*{arg\,max}_c s_c.
$$

All prompt strategies must call the same evaluation logic.
