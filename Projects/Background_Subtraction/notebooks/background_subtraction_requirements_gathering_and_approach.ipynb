## Requirements and Approach

This notebook translates the project problem into implementation requirements. Its numbered sections are intentionally synchronized with the **15 executable cells** of `Background_Subtraction.ipynb`.

## Global Requirements

- Preserve the validated Image Processing module header/logo format.
- Keep explanatory material outside the implementation notebook.
- Apply identical upstream processing when comparing segmentation strategies.
- Preserve the required mask convention: `0 = tool`, `1 = background`.
- Use repository-relative paths and save reproducible figures/results.
- Keep one numbered requirement section for each executable implementation cell; the count is project-dependent and is **not restricted to 13**.

## 1. Configure the Processing Environment

**Requirement.** Centralize paths, frame lists, output directories, and retained numerical parameters in one executable cell.

**Approach.** Import the numerical/image-processing stack, create output directories, define frame `201` as background, define frames `211–291` as evaluation frames, and expose every retained hyperparameter as a named constant.

## 2. Define Reusable Processing and Evaluation Functions

**Requirement.** No downstream cell should duplicate core processing logic.

**Approach.** Implement reusable functions for loading, ground-truth union construction, percentile normalization, spatial Gaussian filtering, Fourier high-pass filtering, feature construction, three segmentation methods, morphology, metrics, and overlays.

## 3. Validate the Dataset and Annotations

**Requirement.** Every frame must have both annotation files and all images must share one shape.

**Approach.** Build a dataset inventory table and enforce completeness/shape assertions before any experiment proceeds.

## 4. Build the Static Background Reference

**Requirement.** Use one fixed processed background for all compared frames.

**Approach.** Load frame 201, apply the common preprocessing function, retain the result as `background_reference`, and save the original/processed diagnostic figure.

## 5. Inspect the Intermediate Pipeline on a Representative Frame

**Requirement.** Make the feature-construction chain observable.

**Approach.** Run frame 251 through preprocessing, subtraction, FFT-domain filtering, morphology, and normalization; visualize the six principal intermediate states.

## 6. Generate the Three Candidate Segmentation Masks

**Requirement.** Compare segmentation strategies under identical upstream conditions.

**Approach.** Apply fixed threshold, Otsu, and two-component GMM to the same representative feature image and finalize all masks with the same morphology/convention.

## 7. Compare the Candidate Strategies Qualitatively

**Requirement.** Provide visual evidence of agreement and error structure.

**Approach.** Place each predicted mask beside its semantic validation overlay using the same representative frame and manual ground truth.

## 8. Evaluate All Strategies Across the Sequence

**Requirement.** Every strategy must be evaluated on every evaluation frame with the same metrics.

**Approach.** Loop over frames `211–291`, cache outputs, compute SAD/MSE/PSNR and Dice/IoU, and assemble one long-form results table.

## 9. Aggregate Metrics and Select the Retained Strategy

**Requirement.** Strategy retention must be data-driven and reproducible.

**Approach.** Group results by strategy, compute mean/std metrics, sort by mean MSE, and assign the top row as `best_strategy`.

## 10. Plot Temporal Validation Curves

**Requirement.** Sequence averages must not hide unstable frames.

**Approach.** Plot SAD, MSE, and PSNR versus frame number for all strategies on common axes.

## 11. Summarize Strategy-Level Performance

**Requirement.** Provide a compact cross-strategy comparison.

**Approach.** Plot mean MSE, mean PSNR, and mean Dice from the aggregated results table.

## 12. Assemble the Final Retained Pipeline

**Requirement.** The final implementation must be callable through a single consistent interface.

**Approach.** Define a strategy dispatcher and `final_pipeline(frame_number, strategy_name=best_strategy)` returning image, feature, prediction, and ground truth.

## 13. Generate the Final Guidance Gallery

**Requirement.** Demonstrate final behavior over the complete evaluation sequence.

**Approach.** Apply the retained pipeline to all evaluation frames and overlay the predicted tool support on contrast-enhanced fluoroscopy.

## 14. Perform Final Qualitative Validation

**Requirement.** Inspect representative temporal positions with explicit error semantics.

**Approach.** For frames 211, 251, and 291, show prediction, manual annotation, and overlap/error overlay side by side.

## 15. Export Results and Verify Deliverables

**Requirement.** Preserve reproducible numerical evidence and confirm figure generation.

**Approach.** Export per-frame metrics and strategy summaries to CSV and print the generated PNG inventory.
