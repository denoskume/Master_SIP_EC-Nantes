## Requirements and Approach

This document defines the engineering requirements for radiometric and geometric image transformations.

The design separates pointwise intensity mappings from spatial coordinate transformations and makes numerical precision, coordinate conventions, inverse mapping, interpolation, boundary behavior, and output geometry explicit. Each transformation must be validated against its analytical model, known geometric behavior, and representative visual evidence.

## Global Requirements

- Use repository-relative paths.
- Keep inputs in `data/` and generated figures in `outputs/figures/`.
- Perform intensity arithmetic safely in floating point before clipping/conversion.
- State coordinate conventions explicitly.
- Use inverse mapping/interpolation intentionally for geometric warps.
- Validate output shape, range, dtype, matrix behavior, and representative known points.
