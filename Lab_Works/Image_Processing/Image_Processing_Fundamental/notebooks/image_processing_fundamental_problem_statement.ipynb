## Context

Digital image processing operates on sampled, quantized arrays whose numerical representation directly affects every downstream operation. Image dimensions, channel order, data type, dynamic range, coordinate convention, and encoding format are therefore part of the computational model rather than incidental implementation details.

This laboratory establishes a controlled numerical framework for inspecting image data, manipulating pixels and regions, characterizing intensity distributions, simulating common noise processes, and comparing images using reproducible quantitative measures.

## Problem Statement

Characterize and validate the numerical representation of binary, grayscale, and RGB images and establish a reproducible set of core image-processing operations.

The work must:

- verify image dimensions, channel organization, dtype, bit depth, and dynamic range;
- manipulate individual pixels, neighborhoods, and regions of interest safely;
- analyze RGB/BGR conventions and grayscale conversion;
- compute descriptive statistics and intensity histograms;
- perform range normalization without overflow or unintended clipping;
- simulate representative Gaussian, salt-and-pepper, Poisson, and speckle noise;
- compare images using MAE, MSE, RMSE, and PSNR;
- assess the numerical consequences of lossless and lossy image encoding;
- validate generated arrays, metrics, and figures.

Geometric transformation, spatial/frequency filtering, segmentation, and projective vision are handled in dedicated modules and are outside this laboratory's scope.

## 20. Image Comparison Metrics

For clean reference $I$ and degraded image $J$, compute

$$
\mathrm{MAE}
=
\frac{1}{N}
\sum_i|I_i-J_i|,
$$

$$
\mathrm{MSE}
=
\frac{1}{N}
\sum_i(I_i-J_i)^2,
$$

$$
\mathrm{RMSE}
=
\sqrt{\mathrm{MSE}},
$$

and, for peak value $L$,

$$
\mathrm{PSNR}
=
20\log_{10}
\left(
\frac{L}{\mathrm{RMSE}}
\right).
$$

Evaluate these metrics for each noise model and present them in one table.

Handle the zero-MSE case explicitly.
