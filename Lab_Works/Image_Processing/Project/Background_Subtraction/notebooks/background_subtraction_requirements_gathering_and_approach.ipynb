## Requirements and Approach

This document defines the experimental and engineering requirements for neurointerventional background subtraction and moving-tool enhancement.

The methodology is organized as a controlled ablation study: background modeling, radiometric normalization, spatial filtering, spectral filtering, morphology, segmentation, and field-of-view restriction are evaluated independently before inclusion in the retained pipeline. Every accepted processing decision must improve or preserve quantitative performance without violating annotation coverage, mask semantics, or reproducibility constraints.

## Global Requirements

- Use repository-relative paths only.
- Preserve the original laboratory mask convention: `0 = tool`, `1 = background`.
- Keep experiments deterministic wherever randomness is involved.
- Change one processing decision at a time during ablation.
- Evaluate every retained change quantitatively and visually.
- Do not retain a more complex method merely because it is more sophisticated.
- Keep all final parameters explicit and reproducible.
- Store generated diagnostics under `outputs/figures/`.

## 1. Validate Data and Ground-Truth Paths

**Requirement.** All 10 frame files and 20 annotation files must be discovered from `data/catheter/`.

**Acceptance criterion.** Fail immediately on missing files, inconsistent shapes, or empty image data.

## 2. Reproduce the Original Baseline Pipeline

**Requirement.** The original first-frame reference method must remain reproducible before improvements are evaluated.

**Acceptance criterion.** Baseline metrics must be computed using the same mask convention and metric definitions used later.

## 3. Replace the First-Frame Background with a Temporal Median

**Requirement.** Temporal median is the only changed variable in this task.

**Acceptance criterion.** Comparison must use the same nine baseline-compatible frames, with frame 201 reported separately.

## 4. Stabilize the Pre-Subtraction Histogram Transformation

**Requirement.** One fixed low/high intensity mapping must be estimated once and reused on every frame.

**Acceptance criterion.** No per-frame intensity remapping may be introduced in the comparison branch.

## 5. Optimize Spatial Gaussian Filtering

**Requirement.** Spatial Gaussian sigma must be swept under otherwise identical conditions.

**Acceptance criterion.** The background reference must be rebuilt in the same Gaussian domain for each candidate sigma.

## 6. Add and Tune Spectral-Domain High-Pass Filtering

**Requirement.** The spectral operation must be explicit in the Fourier domain and controlled by a reproducible cutoff.

**Acceptance criterion.** Only the cutoff/filter decision may vary during this ablation.

## 7. Optimize Morphological Refinement

**Requirement.** Morphological parameter candidates must be evaluated with all upstream retained parameters fixed.

**Acceptance criterion.** Selected morphology must not be justified from appearance alone; metrics and tool preservation are required.

## 8. Optimize the Segmentation Threshold

**Requirement.** Threshold candidates must be evaluated over the complete sequence.

**Acceptance criterion.** The final threshold must come from documented coarse/fine sensitivity evaluation.

## 9. Compare Threshold Segmentation with EM/GMM

**Requirement.** The GMM must use two components and deterministic random state/sample selection.

**Acceptance criterion.** The higher-response component is interpreted as tool; comparison must use exactly the same pre-segmentation feature image as thresholding.

## 10. Restrict Processing to a Valid Field-of-View Mask

**Requirement.** FOV candidates must be tested against raw annotated tool support.

**Acceptance criterion.** A retained FOV must preserve effectively 100% of annotated tool pixels on every evaluated frame.

## 11. Assemble the Final Retained Pipeline

**Requirement.** The final pipeline must expose all retained parameters in one explicit configuration.

**Acceptance criterion.** Rejected experiments must not silently alter the retained branch.

## 12. Compute Sequence-Level Quantitative Evaluation

**Requirement.** SAD, MSE, and PSNR must be produced per frame and summarized by mean and standard deviation.

**Acceptance criterion.** All metric arrays must have one value per evaluated frame and contain finite values.

## 13. Run Numerical and Output-file Validation Checks

**Requirement.** Validation must check data, masks, metrics, retained parameters, FOV safety, and the 17 expected diagnostic figures.

**Acceptance criterion.** Successful execution must finish with `All Background Subtraction validation checks passed.`
