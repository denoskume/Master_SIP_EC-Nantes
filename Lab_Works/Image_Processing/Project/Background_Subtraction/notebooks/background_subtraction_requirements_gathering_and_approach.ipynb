## Requirements and Approach
This document translates the executable background-subtraction notebook into explicit engineering requirements and acceptance criteria. Its section order follows the code notebook exactly: setup is handled globally, followed by the same 14 numbered execution sections.


## Global Requirements
- Use repository-relative paths only.
- Preserve deterministic behavior wherever randomness is used.
- Use the same processing logic for every evaluation frame.
- Preserve the laboratory mask convention: **0 = tool, 1 = background**.
- Keep spatial, spectral, morphological, and segmentation parameters explicit.
- Evaluate predictions against the supplied guidewire + microcatheter annotations.
- Preserve executed outputs and save generated figures under `../outputs/figures/`.
- Export quantitative results under `../outputs/`.


## 1. Reusable Processing Functions
**Requirement.** All reusable processing, segmentation, metric, and visualization operations must be defined once and reused throughout the notebook.

**Acceptance criterion.** No evaluation frame may use a different hidden preprocessing or metric implementation.


## 2. Data and Ground-Truth Validation
**Requirement.** All 10 fluoroscopy frames and their two annotation files must exist and share one image shape.

**Acceptance criterion.** The dataset table reports complete files for every frame and the shape-consistency assertion passes.


## 3. Static Background Reference
**Requirement.** Frame 201 must be processed with the common preprocessing routine and used as the static reference.

**Acceptance criterion.** The raw frame and processed reference are both displayed and the background-reference figure is saved.


## 4. Intermediate Processing Pipeline
**Requirement.** The representative frame must expose the principal intermediate stages of the retained pipeline.

**Acceptance criterion.** Input, preprocessed frame, signed residual, Fourier magnitude, Gaussian high-pass transfer function, and final segmentation feature are visualized together.


## 5. Segmentation Strategy Generation
**Requirement.** Fixed threshold, Otsu, and two-component EM/GMM must operate on the same representative feature image.

**Acceptance criterion.** All three finalized masks follow 0=tool and 1=background; the Otsu threshold and GMM means are reported.


## 6. Representative Strategy Comparison
**Requirement.** Candidate segmentation strategies must be compared under the same representative frame and annotation.

**Acceptance criterion.** A common comparison figure shows each predicted mask and its semantic agreement overlay.


## 7. Sequence-Level Evaluation
**Requirement.** Every strategy must be evaluated on all nine non-background frames.

**Acceptance criterion.** The result table contains one row per frame-strategy pair with finite SAD, MSE, PSNR, Dice, and IoU values.


## 8. Aggregate Metrics and Strategy Selection
**Requirement.** Sequence-level statistics must be computed consistently for every strategy.

**Acceptance criterion.** Mean and standard deviation are reported for all metrics, and the retained strategy is selected from minimum mean MSE exactly as implemented.


## 9. Temporal Metric Curves
**Requirement.** Temporal behavior of the primary laboratory metrics must be visualized.

**Acceptance criterion.** SAD, MSE, and PSNR are plotted versus frame number for all strategies in one saved figure.


## 10. Strategy Summary
**Requirement.** A compact aggregate comparison must complement the temporal curves.

**Acceptance criterion.** Mean MSE, mean PSNR, and mean Dice are displayed for each strategy using the same aggregated results table.


## 11. Final Pipeline Assembly
**Requirement.** The selected method must be available as a reusable final-pipeline function.

**Acceptance criterion.** One function returns the frame, feature image, final prediction, and ground truth for any valid frame number.


## 12. Final Guidance Gallery
**Requirement.** The retained pipeline must produce guidance-ready visual outputs across the complete evaluation sequence.

**Acceptance criterion.** All nine evaluation frames appear in the final guidance gallery using the retained strategy.


## 13. Final Qualitative Validation
**Requirement.** Final qualitative validation must include early, middle, and late sequence examples.

**Acceptance criterion.** Frames 211, 251, and 291 each show prediction, manual annotation, and semantic error overlay.


## 14. Export Results and Output Inventory
**Requirement.** Quantitative results and visual artifacts must be persisted and discoverable.

**Acceptance criterion.** The detailed and summary CSV files are written successfully and the final cell prints the generated figure inventory.
