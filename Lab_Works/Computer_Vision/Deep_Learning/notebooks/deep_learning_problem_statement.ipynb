## Context

This laboratory studies supervised handwritten-digit classification with a one-hidden-layer Multilayer Perceptron (MLP) implemented in PyTorch. The MNIST dataset provides 60,000 training images and 10,000 test images, each representing one digit from 0 to 9 as a $28\times28$ grayscale image.

## Problem Statement

Build, train, compare, and validate three MLP classifiers for MNIST using hidden-layer widths of 128, 256, and 512 neurons.

The complete solution must preserve the laboratory architecture and optimization settings, expose prediction confidence, compare training behavior and test performance, perform confidence-threshold analysis, generate the required figures, and finish with explicit numerical/output validation.
