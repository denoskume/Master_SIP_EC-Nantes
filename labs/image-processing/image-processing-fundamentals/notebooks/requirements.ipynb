## 2. Required Data

This lab does not process an arbitrary image folder: `main.ipynb` assigns five specific reference images to named experiments such as RGB analysis, ROI extraction, histograms, and noise simulation.

The requirement therefore validates the same semantic image set instead of accepting unrelated files.

````python
# Resolve the lab root exactly as the implementation does.
def find_lab_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (candidate / "data").is_dir() and (candidate / "notebooks").is_dir():
            return candidate
    raise FileNotFoundError("Could not locate the image-fundamentals lab root.")


LAB_DIR = find_lab_root(Path.cwd())
DATA_DIR = LAB_DIR / "data"

# Keep the same semantic names used throughout main.ipynb.
IMAGE_FILES = {
    "einstein": DATA_DIR / "einstein.png",
    "peppers": DATA_DIR / "peppers.png",
    "ballons": DATA_DIR / "ballons.jpg",
    "grass": DATA_DIR / "grass.jpg",
    "tower": DATA_DIR / "Elizabeth_Tower_London.jpg",
}

missing = [name for name, path in IMAGE_FILES.items() if not path.is_file()]
assert not missing, f"Missing reference images: {missing}"

print(f"Reference images found: {len(IMAGE_FILES)}")
````
