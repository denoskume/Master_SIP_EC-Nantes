## 2. Required Data

The implementation uses three dataset groups for different experiments: `Fourier/` for spectra and filtering, `PhaseMag/` for phase/magnitude reconstruction, and `Frequency/` for periodic-noise and illumination correction.

The check below validates exactly the files referenced by `main.ipynb`.

````python
# Locate the same lab structure required by main.ipynb.
def find_lab_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (
            (candidate / "data" / "Fourier").is_dir()
            and (candidate / "notebooks" / "main.ipynb").is_file()
        ):
            return candidate
    raise FileNotFoundError("Could not locate the frequency-domain lab root.")


LAB_ROOT = find_lab_root(Path.cwd())
DATA_DIR = LAB_ROOT / "data"

# Group inputs by their actual role in the implementation.
REQUIRED_DATA = {
    "Fourier": [
        "house.png",
        "squares.png",
        "textures.jpg",
        "tiled.png",
        "zebra-wall.png",
    ],
    "PhaseMag": [
        "cat.jpg",
        "wolf.jpg",
    ],
    "Frequency": [
        "astronaut-interference.tif",
        "car-moire-pattern.tif",
        "text-spotshade.tif",
    ],
}

required_paths = [
    DATA_DIR / folder / filename
    for folder, filenames in REQUIRED_DATA.items()
    for filename in filenames
]

missing = [str(path.relative_to(LAB_ROOT)) for path in required_paths if not path.is_file()]
assert not missing, f"Missing frequency-domain inputs: {missing}"

# Pillow is the loader used by main.ipynb, so verify that every source can be opened.
for path in required_paths:
    with Image.open(path) as image:
        image.verify()

print(f"Required source images found: {len(required_paths)}")
````
