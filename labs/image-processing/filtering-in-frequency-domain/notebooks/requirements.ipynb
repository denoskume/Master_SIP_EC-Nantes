## Required packages


````python
packages = [["numpy","numpy"],["matplotlib","matplotlib"],["Pillow","PIL"],["ipykernel","ipykernel"]]

for package_name, import_name in packages:
    try:
        module = __import__(import_name)
        version = getattr(module, "__version__", "unknown")
        print(f"{package_name}: {version}")
    except ImportError:
        print(f"{package_name}: NOT INSTALLED")
````

## Required data files


````python
from pathlib import Path

cwd = Path.cwd().resolve()
lab_root = cwd if (cwd / "data").exists() else cwd.parent
data_dir = lab_root / "data"

data_files = sorted(p for p in data_dir.rglob("*") if p.is_file()) if data_dir.exists() else []

print(f"Data directory: {data_dir}")
print(f"Files found: {len(data_files)}")
for path in data_files:
    print(f"{path.relative_to(lab_root)}: OK")

if not data_files:
    print("No data files found.")
````
