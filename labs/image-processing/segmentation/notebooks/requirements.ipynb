## 2. Required Data

Each source has a distinct role in `main.ipynb`: `hand.png` supports grayscale thresholding and morphology, while `tower.jpg` and `peppers.png` support RGB/color and edge-based segmentation.

The requirement checks exactly these three benchmark inputs.

````python
# Locate the lab with the same hand.png sentinel used by main.ipynb.
def find_lab_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (
            (candidate / "data" / "hand.png").is_file()
            and (candidate / "notebooks" / "main.ipynb").is_file()
        ):
            return candidate
    raise FileNotFoundError("Could not locate the segmentation lab root.")


LAB_ROOT = find_lab_root(Path.cwd())
DATA_DIR = LAB_ROOT / "data"

# These are the only three external images loaded by the implementation.
INPUT_IMAGES = {
    "hand": DATA_DIR / "hand.png",
    "tower": DATA_DIR / "tower.jpg",
    "peppers": DATA_DIR / "peppers.png",
}

missing = [name for name, path in INPUT_IMAGES.items() if not path.is_file()]
assert not missing, f"Missing segmentation inputs: {missing}"

print(f"Segmentation inputs found: {len(INPUT_IMAGES)}")
````
