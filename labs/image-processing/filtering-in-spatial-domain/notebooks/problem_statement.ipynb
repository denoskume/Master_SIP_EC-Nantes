## Context

Spatial-domain filtering modifies an image through local neighborhood operators. The resulting behavior depends jointly on kernel definition, normalization, boundary conditions, numerical precision, image noise characteristics, and the scale of relevant structures.

The laboratory addresses both restoration-oriented filters and derivative/sharpening operators, with emphasis on controlled comparison between filtering objectives such as noise suppression, edge preservation, detail enhancement, and gradient estimation.

## Problem Statement

Design, implement, and evaluate spatial-domain filtering operations for smoothing, denoising, sharpening, and derivative estimation under explicit numerical and boundary conditions.

The work must:

- distinguish correlation from convolution and validate a direct reference implementation;
- evaluate border-handling strategies;
- compare mean and Gaussian smoothing across kernel scales;
- match denoising methods to different noise models;
- evaluate median and bilateral filtering for edge preservation;
- quantify restoration performance using MAE, MSE, RMSE, and PSNR where a reference image exists;
- implement Laplacian sharpening, unsharp masking, and high-boost filtering;
- compare Sobel, Prewitt, and Scharr derivative operators;
- analyze gradient magnitude and orientation;
- verify RGB filtering behavior, numerical safety, and generated diagnostics.

Frequency-domain filtering, segmentation, and learned convolutional models are outside scope.

## 2. Spatial Filtering Formulation

$$
g(x,y)
=
\sum_{i}\sum_{j}
h(i,j)f(x-i,y-j).
$$

State clearly when the expression represents convolution versus correlation.

For every later filter, document:

- kernel support;
- coefficient normalization;
- boundary convention;
- output dtype/range.

Use one common notation throughout the laboratory.

## 12. Quantitative Denoising Metrics

For clean image $I$ and estimate $\hat I$, compute:

$$
\mathrm{MAE}
=
\frac1N\sum_i|I_i-\hat I_i|,
$$

$$
\mathrm{MSE}
=
\frac1N\sum_i(I_i-\hat I_i)^2,
$$

$$
\mathrm{RMSE}=\sqrt{\mathrm{MSE}},
$$

$$
\mathrm{PSNR}
=
20\log_{10}
\left(
\frac{L}{\mathrm{RMSE}}
\right).
$$

Create one comparison table covering the principal denoising configurations.

Handle the zero-error case explicitly.
