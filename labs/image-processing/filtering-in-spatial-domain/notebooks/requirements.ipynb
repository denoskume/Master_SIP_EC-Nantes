## 2. Required Data

The implementation combines two standalone grayscale images with the bilateral-filter dataset. Einstein provides reference/noisy pairs, while Taj Mahal, zebra, and monarch provide RGB examples for edge-preserving filtering.

The check below follows the exact paths later loaded by `main.ipynb`.

````python
# Resolve the lab root using the same data/notebooks structure as main.ipynb.
def find_lab_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (candidate / "data").is_dir() and (candidate / "notebooks").is_dir():
            return candidate
    raise FileNotFoundError("Could not locate the spatial-filtering lab root.")


LAB_DIR = find_lab_root(Path.cwd())
DATA_DIR = LAB_DIR / "data"

EINSTEIN_DIR = DATA_DIR / "bilateral" / "einstein"
TAJ_DIR = DATA_DIR / "bilateral" / "tajMahal"
ZEBRA_DIR = DATA_DIR / "bilateral" / "zebra"
MONARCH_DIR = DATA_DIR / "bilateral" / "monarch"

# Every file below is loaded later by main.ipynb.
required_paths = [
    DATA_DIR / "moon-blurred.tif",
    DATA_DIR / "ascentB.png",
    EINSTEIN_DIR / "ref.png",
    EINSTEIN_DIR / "einstein_gaussian_0.png",
    EINSTEIN_DIR / "einstein_saltAndPepper.png",
    EINSTEIN_DIR / "einstein_speckle_0.png",
    TAJ_DIR / "ref.jpg",
    ZEBRA_DIR / "ref.jpg",
    MONARCH_DIR / "ref.png",
]

missing = [str(path.relative_to(LAB_DIR)) for path in required_paths if not path.is_file()]
assert not missing, f"Missing spatial-filtering inputs: {missing}"

print(f"Required source images found: {len(required_paths)}")
````
