## Context

Image transformation covers two distinct classes of operations: radiometric mappings that modify pixel intensities and geometric mappings that modify spatial coordinates. Both require explicit numerical models because clipping, interpolation, coordinate origin, transformation order, and resampling strategy can materially alter the result.

The laboratory treats each transformation as a parameterized operator whose behavior can be checked against analytical expectations and known geometric relationships.

## Problem Statement

Implement and validate a set of radiometric and 2-D geometric image transformations with explicit control of numerical precision, coordinate conventions, interpolation, and output geometry.

The work must:

- implement negative, brightness, contrast, stretching, logarithmic, gamma, and histogram-equalization mappings;
- formulate translation, rotation, scaling, reflection, shear, and affine transforms in homogeneous coordinates;
- distinguish transformations about the image origin from transformations about the image center;
- compare forward and inverse mapping;
- evaluate nearest-neighbor, bilinear, and bicubic interpolation;
- demonstrate transformation composition and order dependence;
- verify array shape, intensity range, dtype behavior, and representative transformed coordinates;
- generate quantitative and visual diagnostics for the retained implementations.

Spatial filtering, frequency-domain processing, segmentation, and projective homography estimation are outside scope.

## 3. Intensity Transformation Model

Model pointwise radiometric transformation as

$$
s=T(r),
$$

where $r$ is the input intensity and $s$ is the output intensity.

For every intensity transform used later:

1. state $T(r)$ analytically;
2. identify its valid input/output range;
3. perform arithmetic in floating point;
4. clip only when required by the target representation;
5. verify monotonicity where theoretically expected.

Distinguish pointwise intensity mappings from geometric coordinate transformations.

## 4. Image Negative

For an $L$-level image, implement

$$
s=(L-1)-r.
$$

For 8-bit data, use $L=256$.

Apply the transform to representative grayscale and RGB images.

Verify numerically that

$$
r+s=255
$$

for valid 8-bit scalar intensities.

Compare the original and negative histograms and confirm that intensity ordering is reversed around the center of the range.

Save the result as `02_negative.png`.

## 5. Brightness and Contrast

Implement the affine intensity mapping

$$
s=ar+b,
$$

where $a$ controls contrast and $b$ controls brightness.

Evaluate at least:

- $a<1$ and $a>1$;
- positive and negative $b$.

Perform the calculation in floating point and clip safely to the display range.

Save a comparison figure as `03_brightness_contrast.png`.

Report the fraction of pixels clipped for each tested configuration.

## 6. Contrast Stretching

Implement min-max contrast stretching:

$$
s=
\frac{r-r_{min}}

## 7. Logarithmic Transformation

$$

$$

with $c$ chosen so that the maximum input maps to the desired output range.

Save the result as `05_log_transform.png`.

## 8. Gamma / Power-Law Transformation

## 13. Fundamental Geometric Transformation Matrices

Derive and implement homogeneous matrices for:

Translation:

$$
T=
\begin{bmatrix}
1&0&t_x\\
0&1&t_y\\
0&0&1
\end{bmatrix},
$$

Scaling:

$$
S=
\begin{bmatrix}
s_x&0&0\\
0&s_y&0\\
0&0&1
\end{bmatrix},
$$

Rotation:

$$
R=
\begin{bmatrix}
\cos\theta&-\sin\theta&0\\
\sin\theta&\cos\theta&0\\
0&0&1
\end{bmatrix}.
$$

Also define reflection and shear matrices.

Validate every matrix on a small set of known points before applying it to an image.
