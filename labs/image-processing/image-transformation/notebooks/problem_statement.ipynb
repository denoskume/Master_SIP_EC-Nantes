## Context

Image transformation covers two distinct classes of operations: radiometric mappings that modify pixel intensities and geometric mappings that modify spatial coordinates. Both require explicit numerical models because clipping, interpolation, coordinate origin, transformation order, and resampling strategy can materially alter the result.

The laboratory treats each transformation as a parameterized operator whose behavior can be checked against analytical expectations and known geometric relationships.

**Problem.** Implement and validate a set of radiometric and 2-D geometric image transformations with explicit control of numerical precision, coordinate conventions, interpolation, and output geometry.

The work must:

- implement negative, brightness, contrast, stretching, logarithmic, gamma, and histogram-equalization mappings;
- formulate translation, rotation, scaling, reflection, shear, and affine transforms in homogeneous coordinates;
- distinguish transformations about the image origin from transformations about the image center;
- compare forward and inverse mapping;
- evaluate nearest-neighbor, bilinear, and bicubic interpolation;
- demonstrate transformation composition and order dependence;
- verify array shape, intensity range, dtype behavior, and representative transformed coordinates;
- generate quantitative and visual diagnostics for the retained implementations.

Spatial filtering, frequency-domain processing, segmentation, and projective homography estimation are outside scope.
