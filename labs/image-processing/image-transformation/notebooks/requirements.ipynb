## 2. Required Data

The transformation notebook uses five named reference images repeatedly in RGB and grayscale form. Because the implementation addresses them by semantic name, the requirement must validate those exact inputs rather than any image found in `data/`.

````python
# Resolve the lab root using the same data/notebooks contract as main.ipynb.
def find_lab_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (candidate / "data").is_dir() and (candidate / "notebooks").is_dir():
            return candidate
    raise FileNotFoundError("Could not locate the image-transformation lab root.")


LAB_DIR = find_lab_root(Path.cwd())
DATA_DIR = LAB_DIR / "data"

# Mirror the exact IMAGE_FILES mapping consumed by main.ipynb.
IMAGE_FILES = {
    "ascent": DATA_DIR / "ascentB.png",
    "ballons": DATA_DIR / "ballons.jpg",
    "einstein": DATA_DIR / "einstein.png",
    "tower": DATA_DIR / "Elizabeth_Tower_London.jpg",
    "peppers": DATA_DIR / "peppers.png",
}

missing = [name for name, path in IMAGE_FILES.items() if not path.is_file()]
assert not missing, f"Missing transformation inputs: {missing}"

print(f"Reference images found: {len(IMAGE_FILES)}")
````
